## 3. Data Preprocessing

In this step, we select the relevant features for our clustering analysis: 'Annual Income (k$)' and 'Spending Score (1-100)'. These two features form our feature matrix `X`, which will be used as input for the clustering algorithms.

```python
# We will use 'Annual Income (k$)' and 'Spending Score (1-100)' as our base features for standard clustering comparison
# We apply standard scaling because scaling is essential when applying distance-based algorithms like K-Means

base_features = ['Annual Income (k$)', 'Spending Score (1-100)']
X_raw = dataset[base_features].values

scaler = StandardScaler()
X = scaler.fit_transform(X_raw)

print("Features selected and scaled successfully.")
print("Base features shape:", X.shape)
```

### Graph 2: Elbow Method for K-Means

This chart plots the Within-Cluster Sum of Squares (WCSS) against different cluster numbers to find the 'elbow' point where adding more clusters yields diminishing returns. It is calculated on a robust sample of 50,000 data points to optimize computational resources and time.

```python
# Safeguarded sampling limits for WCSS elbow plot
elbow_sample_size = min(50000, X.shape[0])
np.random.seed(42)
X_elbow_sample = X[np.random.choice(X.shape[0], elbow_sample_size, replace=False)]

wcss = []
for i in range(1, 11):
    kmeans_model = KMeans(n_clusters=i, init='k-means++', random_state=42, n_init=10)
    kmeans_model.fit(X_elbow_sample)
    wcss.append(kmeans_model.inertia_)

plt.figure(figsize=(10, 6))
plt.plot(range(1, 11), wcss, marker='o', linestyle='--', color='darkblue')
plt.title(f'2. Elbow Method for Optimal K ({elbow_sample_size:,} Samples)')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS')
plt.grid(True, linestyle='--', alpha=0.6)
plt.xticks(np.arange(1, 11, 1))
plt.show()
```

## 5. K-Means Elbow Method

The Elbow Method is used to determine the optimal number of clusters (K) for K-Means clustering. It plots the Within-Cluster Sum of Squares (WCSS) against the number of clusters. The 'elbow' point in the graph, where the rate of decrease in WCSS significantly slows down, suggests an optimal K value.

```python
# Safeguarded sampling limits
elbow_sample_size = min(50000, X.shape[0])
np.random.seed(42)
X_elbow_sample = X[np.random.choice(X.shape[0], elbow_sample_size, replace=False)]

wcss = []
for i in range(1, 11):
    kmeans_model = KMeans(n_clusters=i, init='k-means++', random_state=42, n_init=10)
    kmeans_model.fit(X_elbow_sample)
    wcss.append(kmeans_model.inertia_)

plt.figure(figsize=(10, 6))
plt.plot(range(1, 11), wcss, marker='o', linestyle='--')
plt.title(f'Elbow Method for Optimal K ({elbow_sample_size:,} Samples)')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS')
plt.grid(True, linestyle='--', alpha=0.6)
plt.xticks(np.arange(1, 11, 1))
plt.show()
```

## 6. Final K-Means Model

Based on the Elbow Method (typically, the 'elbow' is observed at K=5 in this dataset), we will now train the K-Means model with 5 clusters. We then generate cluster labels for each customer and print the coordinates of the cluster centroids.

```python
# Train the K-Means model on the FULL 1-million-row scaled dataset
kmeans = KMeans(n_clusters=5, init='k-means++', random_state=42, n_init=10)
y_kmeans = kmeans.fit_predict(X)

# Calculate original cluster centers (inverse transform the scaled centroids)
centroids_raw = scaler.inverse_transform(kmeans.cluster_centers_)

print("K-Means trained successfully on 1,000,000 records!")
print("\nK-Means Cluster Centroids (Raw Units):")
for i, centroid in enumerate(centroids_raw):
    print(f"Cluster {i+1}: Annual Income = {centroid[0]:.2f} k$, Spending Score = {centroid[1]:.2f}")
```

## 9. Customer Segment Interpretation

Here, we dynamically interpret the meaning of each cluster based on its centroid's annual income and spending score. Instead of hard-coding, we analyze the relative positions of centroids to assign business-relevant labels like 'VIP Customer', 'Careful Spender', etc.

```python
def interpret_cluster(income, spending_score):
    if income > 70 and spending_score > 70:
        return "VIP / Target Customer (High Income + High Spending)"
    elif income > 70 and spending_score < 40:
        return "Careful Spender (High Income + Low Spending)"
    elif income < 40 and spending_score < 40:
        return "Sensible Spender (Low Income + Low Spending)"
    elif income < 40 and spending_score > 70:
        return "Careless Spender (Low Income + High Spending)"
    else:
        return "Average Customer (Middle/Average Income + Middle/Average Spending)"

print("--- Customer Segment Interpretation (K-Means) ---")
cluster_data = []
for i, centroid in enumerate(centroids_raw):
    income = centroid[0]
    spending_score = centroid[1]
    segment = interpret_cluster(income, spending_score)
    cluster_data.append([i + 1, f"{income:.2f}", f"{spending_score:.2f}", segment])

interpretation_df = pd.DataFrame(cluster_data, columns=['Cluster Number', 'Income (k$)', 'Spending Score', 'Customer Segment'])
display(interpretation_df)
```

## 10. Silhouette Score

The Silhouette Score is a metric used to evaluate the quality of clustering results. It measures how similar an object is to its own cluster compared to other clusters. A higher Silhouette Score indicates better-defined and more separated clusters. The score ranges from -1 to 1, where 1 signifies excellent clustering, 0 indicates overlapping clusters, and -1 suggests incorrect clustering.

```python
import numpy as np
from sklearn.metrics import silhouette_score

# Ensure we are using the full dataset scaling shapes and predictions
X_eval = scaler.transform(X_raw)
y_eval_kmeans = kmeans.predict(X_eval)

eval_sample_size = min(20000, X_eval.shape[0])
np.random.seed(42)
eval_indices = np.random.choice(X_eval.shape[0], eval_sample_size, replace=False)

sil_score = silhouette_score(X_eval[eval_indices], y_eval_kmeans[eval_indices])
print(f"Silhouette Score for K-Means Clustering (calculated on {eval_sample_size:,} samples): {sil_score:.4f}")
```

## 11. Hierarchical Clustering

Hierarchical clustering builds a hierarchy of clusters. Agglomerative clustering (bottom-up) starts with individual data points as clusters and merges them based on similarity. We use 5 clusters, Euclidean distance as the metric, and 'ward' linkage, which minimizes the variance of the clusters being merged.

```python
# Safeguarded sampling limits
hc_sample_size = min(5000, X.shape[0])
np.random.seed(42)
hc_sample_indices = np.random.choice(X.shape[0], hc_sample_size, replace=False)
X_hc_sample = X[hc_sample_indices]
X_hc_sample_raw = X_raw[hc_sample_indices]

hc = AgglomerativeClustering(n_clusters=5, metric='euclidean', linkage='ward')
y_hc = hc.fit_predict(X_hc_sample)

print(f"Hierarchical Clustering completed successfully on a representative sample of {hc_sample_size:,} records.")
```
